```python
import pickle
import pandas as pd
```

```python
with open('model.bin', 'rb') as f_in:
    dv, model = pickle.load(f_in)
```

```python
categorical = ['PULocationID', 'DOLocationID']

def read_data(filename):
    df = pd.read_parquet(filename)
    
    df['duration'] = df.tpep_dropoff_datetime - df.tpep_pickup_datetime
    df['duration'] = df.duration.dt.total_seconds() / 60

    df = df[(df.duration >= 1) & (df.duration <= 60)].copy()

    df[categorical] = df[categorical].fillna(-1).astype('int').astype('str')
    
    return df
```

```python
df = read_data('https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2023-03.parquet')
```

```python
dicts = df[categorical].to_dict(orient='records')
X_val = dv.transform(dicts)
y_pred = model.predict(X_val)
```

```python
import numpy as np
print(f'Standard deviation of predicted duration: {np.std(y_pred):.4f}')
```

```python
# Create the ride_id column
year = 2023
month = 3
df['ride_id'] = f'{year:04d}/{month:02d}_' + df.index.astype('str')

# Create a new dataframe with just ride_id and predictions
df_result = pd.DataFrame({
    'ride_id': df['ride_id'],
    'predicted_duration': y_pred
})

# Save to parquet
output_file = 'output_homework_4.parquet'
df_result.to_parquet(
    output_file,
    engine='pyarrow',
    compression=None,
    index=False
)
```

```python
import os

# Get file size in bytes
file_size = os.path.getsize('output_homework_4.parquet')

# Convert to megabytes
file_size_mb = file_size / (1024 * 1024)

print(f'File size: {file_size_mb:.2f} MB')
```
